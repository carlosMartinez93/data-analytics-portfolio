```python

import pandas as pd
import numpy as np
```


## 1. Load datasets (English schema)


```python

patients = pd.read_excel("Bases/Bases Iniciais/Patients_EN.xlsx")
beds = pd.read_excel("Bases/Bases Iniciais/F_Hospital_Bed_EN.xlsx")
events = pd.read_csv("Bases/Bases Iniciais/Hospital_Events_EN.csv")


patients.head(), beds.head(), events.head()
```


## 2. Basic Data Validation


```python

print("Patients:", patients.shape)
print("Beds:", beds.shape)
print("Events:", events.shape)

patients.isna().mean().sort_values(ascending=False).head()
```


## 3. Patient Profile Analysis


```python

patients.groupby("Age_Group")["Patient_ID"].nunique().sort_values(ascending=False)
```

```python

pd.crosstab(
    patients["Age_Group"],
    patients["Admission_Type"],
    normalize="index"
).round(3)
```


**Insight:** Elderly patients concentrate a higher ICU admission rate.



## 4. Length of Stay (LOS) Analysis


```python

events["Length_of_Stay_Days"].describe()
```

```python

events.groupby("Admission_Type")["Length_of_Stay_Days"].mean().round(2)
```


**Insight:** ICU stays are significantly longer, increasing sustained bed occupancy.



## 5. Bed Status Snapshot


```python

beds["Bed_Status"].value_counts(normalize=True).round(3)
```
